```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
```

```python
# Reload the report here so this cell works even if run on its own
with open('blood_work.txt', 'r') as f:
    blood_report = f.read()

extraction_prompt = f"""
You are a medical data extraction assistant.

From the blood report below, extract ALL test values and classify each one as a HIGH, LOW, or NORMAL
based on the reference ranges provided in the report.
Format your response as:
- Test Name: value | Status: HIGH|LOW|NORMAL | Reference: Range

Blood Report:
{blood_report}
"""

extraction_response = llm.invoke(extraction_prompt)
extracted_values = extraction_response.content

print("===== STAGE 1: EXTRACTED VALUES =====")
print(extracted_values)
```

```python
diet_prompt = f"""
You are a clinical nutritionist specializing in Indian dietary habits.
s
Based on the blood work analysis below, write:
1. A short health summary in 3 lines explaining the patient's condition in a simple language.
2. A short, practical Indian diet plan having only two sections (1) Food to avoid (2) Food to eat more of.
Do not include any other sections in diet plan.

Blood work analysis:
{extracted_values}
"""

diet_response = llm.invoke(diet_prompt)

print("===== STAGE 2: DIET PLAN =====")
print(diet_response.text)
```
