```python
import collections

import chromadb

client = chromadb.Client()
```

```python
# get_or_create_collection makes this cell safe to re-run:
# create_collection would raise "Collection already exists" on a second run
collection = client.get_or_create_collection("news_articles")
```

```python
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Apple is leading in a smart phone game with iphone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C, and various antioxidants",
        "SpaceX got nasa contract worth 10 billion $",
    ]
)
```

```python
#Fetch documents + embeddings

data = collection.get(
    include=["documents", "embeddings", "metadatas"]  # note: "metadatas" (plural) in chromadb 1.x
)

#Printing embeddings:
for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}:")
    print("Text:", data["documents"][i])
    print("Embedding length:", len(emb))
    print("First 10 values:", emb[:10])
```

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results=2
)

print(results)
```
